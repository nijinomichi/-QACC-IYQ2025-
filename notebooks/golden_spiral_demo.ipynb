This notebook demonstrates how failure logs are transformed into

```python
import math
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np

PHI = (1 + math.sqrt(5)) / 2
EPSILON = 1 / PHI
KINTSUGI_GOLD = '#C9A84C'
BG_COLOR = '#0a0a0f'
```

```python
# Sample failure log — replace with your own
errors = [
    {'id': 'commit-5c45a90', 'severity': 0.3, 'description': 'Wrong repo commit → Kintsugi artifact'},
    {'id': 'perspective-03', 'severity': 0.2, 'description': 'Design Beginning in Failure'},
    {'id': 'perspective-11', 'severity': 0.4, 'description': 'Healer of Fractures'},
    {'id': 'perspective-21', 'severity': 0.5, 'description': 'Score Structure Within Errors'},
    {'id': 'banana-01',      'severity': 0.1, 'description': 'Banana Conference — session open'},
]

def golden_spiral_positions(n):
    positions = []
    for i in range(n):
        angle = i * 2 * math.pi / PHI
        radius = math.sqrt(i + 1) * PHI
        positions.append((radius * math.cos(angle), radius * math.sin(angle)))
    return positions

positions = golden_spiral_positions(len(errors))
trust_coefficient = 1.0 + EPSILON * sum(e['severity'] * PHI for e in errors)
print(f'Trust Coefficient H_R = {trust_coefficient:.4f} (target: φ = {PHI:.4f})')
```

```python
fig, ax = plt.subplots(figsize=(10, 10), facecolor=BG_COLOR)
ax.set_facecolor(BG_COLOR)
ax.set_aspect('equal')
ax.axis('off')

# Draw golden spiral reference
theta = np.linspace(0, 6 * math.pi, 1000)
r = np.sqrt(theta / (2 * math.pi) + 1) * PHI
xs = r * np.cos(theta * 2 * math.pi / PHI)
ys = r * np.sin(theta * 2 * math.pi / PHI)
ax.plot(xs, ys, color=KINTSUGI_GOLD, alpha=0.15, linewidth=1)

# Plot each error as a glowing node
for i, (e, (x, y)) in enumerate(zip(errors, positions)):
    size = e['severity'] * 800 + 100
    ax.scatter(x, y, s=size, color=KINTSUGI_GOLD, alpha=0.7, zorder=5)
    ax.scatter(x, y, s=size * 2.5, color=KINTSUGI_GOLD, alpha=0.08, zorder=4)
    ax.text(x, y + 0.4, e['id'], color='white', fontsize=7,
            ha='center', va='bottom', alpha=0.8)

ax.set_title(
    f'Golden Spiral of Failures — H_R = {trust_coefficient:.4f}',
    color='white', fontsize=13, pad=20
)
plt.tight_layout()
plt.savefig('golden_spiral_output.png', dpi=150, bbox_inches='tight', facecolor=BG_COLOR)
plt.show()
print('Saved: golden_spiral_output.png')
```
